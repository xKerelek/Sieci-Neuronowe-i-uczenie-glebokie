Zadanie 4 -  "Kalkulator operacji macierzowych z walidacją wymiarów"


```python
def kalkulator_macierzy(A: np.array, B: np.array, operacja: str):
  ksztalt_A = A.shape
  ksztalt_B = B.shape

  if(operacja in "dodawanie", "odejmowanie"):
    if ksztalt_A != ksztalt_B:
      print("Wymiary muszą być identyczne")
      return None

    if operacja == "dodawanie":
      wynik = A + B
    else:
      wynik = A - B

  elif operacja == "mnozenie":
    if ksztalt_A[1] != ksztalt_B[0]:
      print("Ilosc kolumn w macierzy A musi byc rowna ilosci wierszy w macierzy B")
      return None
    return A @ B

  else:
    print("Nieznana operacja")
    return None

  return wynik
```

```python
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])
M3 = np.array([[1, 2, 3], [4, 5, 6]])

print('1. Poprawne dodawanie (2x2 + 2x2):')
print(kalkulator_macierzy(M1, M2, 'dodawanie'))

print('\n2. Błędne odejmowanie (2x2 - 2x3):')
kalkulator_macierzy(M1, M3, 'odejmowanie')

print('\n3. Poprawne mnożenie (2x2 * 2x3 -> wynik 2x3):')
print(kalkulator_macierzy(M1, M3, 'mnozenie'))

print('\n4. Błędne mnożenie (2x3 * 2x2):')
kalkulator_macierzy(M3, M1, 'mnozenie')
```

Zadanie 6 - Iloczyn skalarny i podobieństwo wektorów

```python
import time

def iloczyn_skalarny_petla(u, v):
  wynik = 0.0
  for i in range(len(u)):
    wynik += u[i] * v[i]
  return wynik

def iloczyn_skalarny_numpy(u: np.ndarray, v: np.ndarray):
  return np.dot(u, v)


def podobienstwo_kosinusowe(u: np.ndarray, v: np.ndarray):
  licznik = u @ v
  mianownik = np.linalg.norm(u) * np.linalg.norm(v)

  if mianownik == 0:
    return 0.0
  return licznik / mianownik


n = 1_000_000

u = np.random.rand(n)
v = np.random.rand(n)

start_petla = time.perf_counter()
wynik_petla = iloczyn_skalarny_petla(u, v)
czas_petla = time.perf_counter() - start_petla

start_np = time.perf_counter()
wynik_np = iloczyn_skalarny_numpy(u, v)
czas_np = time.perf_counter() - start_np

cos_sim = podobienstwo_kosinusowe(u, v)

print(f"Wymiar wektorów: n = {n}")
print("-" * 50)
print(f"Wynik (pętla for): {wynik_petla:.4f} | Czas: {czas_petla:.6f} s")
print(f"Wynik (NumPy):     {wynik_np:.4f} | Czas: {czas_np:.6f} s")
print(f"NumPy jest około {czas_petla / czas_np:.0f}x szybsze niż pętla for.")
print("-" * 50)
print(f"Podobieństwo kosinusowe cos_sim(u, v): {cos_sim:.6f}")
```

```python
def sigmoid(z: np.ndarray) -> np.ndarray:
  return 1 / (1 + np.exp(-z))

  np.random.seed(42)
  X = np.random.randn(5, 8)
  W = np.random.randn(8, 3)
  b = np.random.randn(3)

  Z = (X @ W) + b
  A = sigmoid(Z)

print("=== WYMIARY TENSORÓW (SHAPE) ===")
print(f"Wejście X:                {X.shape}  -> (liczba_obiektów, liczba_cech)")
print(
    f"Macierz wag W:            {W.shape}  -> (liczba_cech, liczba_neuronów)"
)
print(f"Wektor obciążeń b:        {b.shape}     -> (liczba_neuronów,)")
print(
    f"Przekształcenie liniowe Z:{Z.shape}  -> (liczba_obiektów,"
    " liczba_neuronów)"
)
print(
    f"Wyjście po aktywacji A:   {A.shape}  -> (liczba_obiektów,"
    " liczba_neuronów)"
)

print("\n=== WARTOŚCI WYJŚCIOWE WARSTWY (po Sigmoid) ===")
print(np.round(A, 4))
```

Zadanie 8 - Przygotowanie minibatch obrazów w siecie

```python
from graphviz import Digraph

batch_obrazow = np.random.randint(0, 256, size=(10, 32, 32, 3), dtype=np.uint8)

macierz_mlp = batch_obrazow.reshape(batch_obrazow.shape[0], -1)

ksztalt_wejsciowy = batch_obrazow.shape
ksztalt_wyjsciowy = macierz_mlp.shape

elementy_wejscie = batch_obrazow.size
elementy_wyjscie = macierz_mlp.size
czy_taka_sama_liczba = (elementy_wejscie == elementy_wyjscie)

print("=== PODSUMOWANIE WYMIARÓW ===")
print(f"Kształt tensora wejściowego (4D): {ksztalt_wejsciowy}")
print(f"Kształt macierzy wynikowej (2D):  {ksztalt_wyjsciowy}")
print("-" * 45)
print(f"Liczba elementów przed reshape:   {elementy_wejscie}")
print(f"Liczba elementów po reshape:      {elementy_wyjscie}")
print(f"Czy liczba elementów jest równa?  {czy_taka_sama_liczba}")

dot = Digraph(comment='Przygotowanie minibatcha obrazow do sieci MLP')
dot.attr(rankdir='LR', size='10,4')
dot.attr('node', shape='record', style='filled', fontname='Helvetica')

dot.node('T4D',
         f'{{Tensor wejściowy 4D (Obrazki RGB)|'
         f'Kształt: {ksztalt_wejsciowy}|'
         f'Batch: 10 | Wysokość: 32 | Szerokość: 32 | Kanały: 3|'
         f'Liczba elementów (.size): {elementy_wejscie}}}',
         fillcolor='#E3F2FD')

dot.node('OP',
         '{Operacja spłaszczania|.reshape(10, -1)\nlub .reshape(10, 32*32*3)}',
         shape='Mrecord', fillcolor='#FFF9C4')

dot.node('M2D',
         f'{{Macierz wyjściowa 2D (Wejście do MLP)|'
         f'Kształt: {ksztalt_wyjsciowy}|'
         f'Wiersze (obrazy): 10 | Kolumny (cechy): 3072|'
         f'Liczba elementów (.size): {elementy_wyjscie}}}',
         fillcolor='#E8F5E9')

dot.edge('T4D', 'OP', label=' 10 x (32 x 32 x 3) ')
dot.edge('OP', 'M2D', label=' spłaszczenie (flatten) ')

dot.render('podsumowanie_wymiarow_mlp', format='png', cleanup=True)
print("\nGraf został wygenerowany i zapisany do pliku 'podsumowanie_wymiarow_mlp.png'.")
```
