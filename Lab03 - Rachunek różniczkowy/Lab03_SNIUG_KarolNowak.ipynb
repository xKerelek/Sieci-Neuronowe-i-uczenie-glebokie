Zadanie 1 - Pochodna numeryczna wielomianu wyższego stopnia.

Napisz funkcję numerical diff(func, x, h=1e-5), która oblicza pochodną dowolnej funkcji przekazanej jako argument w zadanym punkcie x. Następnie przetestuj ją na funkcji wielomianowej
trzeciego stopnia f(x) = 2x
3 − 4x
2 + x − 5 w punkcie x = 3.0.

```python
def f(x):
  return 2 * (x**3) - 4 * (x**2) + x - 5

def numerical_diff(func, x, h=1e-5):
  return (func(x + h) - func(x)) / h

x_val = 3.0
slope = numerical_diff(f, x_val)
print(f"Nachylenie w punkcie x={x_val}: {slope}")
```

Zadanie 2 - Pochodna funkcji ReLU

Zaimplementuj funkcję aktywacji ReLU (Rectified Linear Unit) oraz jej pochodną. Pamiętaj, że
pochodna ReLU wynosi 0 dla x < 0 oraz 1 dla x > 0 (w punkcie x = 0 można ją zdefiniować jako
0). Przetestuj jej wartość i pochodną dla punktów x ∈ [−2.0, 0.0, 3.0].

```python
def relu(x):
  return max(0.0, x)

def relu_derivative(x):
  return 1.0 if x > 0 else 0.0

test_points = [-2.0, 0.0, 3.0]
for x_val in test_points:
  print(f"x = {x_val} -> ReLU({x_val}) = {relu(x_val)}, Pochodna = {relu_derivative(x_val)}")
```

Zadanie 3 - Spadek gradientu dla funkcji kosztu z dwiema wagami

Dana jest funkcja kosztu L(w1, w2) = (w1 − 2)2 + (w2 + 3)2 + 4. Pochodne cząstkowe wynoszą
odpowiednio ∂L
∂w1
= 2(w1 − 2) oraz ∂L
∂w2
= 2(w2 + 3). Zaimplementuj 5 kroków algorytmu spadku
gradientu startując z punktu początkowego w1 = 0.0, w2 = 0.0 przy współczynniku uczenia η = 0.2.

```python
def loss_func(w1, w2):
  return (w1 - 2)**2 + (w2 + 3)**2 + 4

w1, w2 = 0.0, 0.0
lr = 0.2

for step in range(5):
  grad_w1 = 2 * (w1 - 2)
  grad_w2 = 2 * (w2 + 3)

  w1 -= lr * grad_w1
  w2 -= lr * grad_w2

  print(f"Krok {step+1}: w1 = {w1:.4f}, w2 = {w2:.4f}, strata = {loss_func(w1, w2):.4f}")
```

Rozważ prosty model składający się z dwóch operacji (warstw): u = w1 · x oraz y = u

względem wagi w1 dla wartości wejściowej x = 2.0 oraz wagi w1 = 1.5.

```python
def calc_u(w1, x):
  return w1 * x

def calc_y(u):
  return u**2

def du_dw1(x):
  return x

def dy_du(u):
  return 2 * u

x = 2.0
w1 = 1.5

u = calc_u(w1, x)
chain_derivative = dy_du(u) * du_dw1(x)
print(f"Wartość pośrednia u: {u}")
print(f"Wynik reguły łańcuchowej (analitycznie) dy/dw1: {chain_derivative}")

h = 1e-5
y_initial = calc_y(calc_u(w1, x))
y_shifted = calc_y(calc_u(w1 + h, x))
numerical_grad = (y_shifted - y_initial) / h
print(f"Wynik pochodnej (numerycznie) dy/dw1: {numerical_grad:.4f}")
```

Zadanie 5 - Wpływ wyboru współczynnika uczenia na zbieżność

Dana jest funkcja kosztu L(w) = w
2
i jej pochodna dL
dw = 2w. Startując z punktu w = 4.0, przeprowadź symulację 3 kroków spadku gradientu dla trzech różnych współczynników uczenia: η1 = 0.1,
η2 = 1.0 oraz η3 = 1.5. Zaobserwuj, w którym przypadku algorytm zbiega do minimum, w którym
oscyluje, a w którym staje się niestabilny (rozbieżny).

```python
def loss_func(w):
  return w**2

w_start = 2.0
learning_rates = [0.1, 1.0, 1.5]

for lr in learning_rates:
    w = w_start
    print(f"--- Symulacja dla LR = {lr} (start w = {w}) ---")
    for step in range(3):
        grad = 2 * w
        w = w - lr * grad
        print(f"Krok {step+1}: w = {w:.4f}, strata = {loss_func(w):.4f}\n")
```
